### Criterio para la columna descuentos en ventas
* Ventas de 2500 5 % de descuento
* Ventas de 1000 3 % de descuento
* Ventas de 750  0.5 % de descuento

```python
def calcular_descuento(subtotal):
    if subtotal >= 2500:
        return subtotal * 0.05
    elif subtotal >= 1000 and subtotal <= 1999:
        return subtotal * 0.03
    elif subtotal >= 750 and subtotal <= 999:
        return subtotal * 0.005
    else:
        return 0
```

```python
df1_cleaned['Discount'] = df1_cleaned['SubTotal'].apply(calcular_descuento)
```

## Estandarización (Normalización Z-Score)

Este método transforma los datos de tal manera que tengan **media 0** y **desviación estándar 1**. Es útil cuando tus datos siguen una distribución normal.

### Fórmula:

$$
x_{\text{estandarizado}} = \frac{x - \mu}{\sigma}
$$

donde:
- $\mu$ es la **media** de la columna.
- $\sigma$ es la **desviación estándar** de la columna.

### Ventajas:
- Reduce el impacto de *outliers*.
- Útil para algoritmos sensibles a la escala, como **K-Means**.

### Desventajas:
- Puede distorsionar la distribución original.
